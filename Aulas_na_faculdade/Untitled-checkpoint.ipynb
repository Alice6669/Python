```python
# Nomes de amigues.
nomes = ["Bruna", "Sora", "Aly", "Isadora", "Jamine", "Larissa", "Luísa", "Maria Flor Ramos", "Maria Flor Amorim", "Aúrea", "Ilana"]
for numero in range(11):
    print("Bom dia, "+ nomes[numero] + ", como está?")
```
